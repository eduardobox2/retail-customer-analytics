```python
import pandas as pd
import sqlite3
import os

# 1. Detectar automáticamente la ruta correcta de la base de datos
db_path = 'data/retail.db' if os.path.exists('data/retail.db') else '../data/retail.db'
conn = sqlite3.connect(db_path)

# 2. Extraer todos los datos a un DataFrame de Pandas
query = "SELECT * FROM retail_sales"
df = pd.read_sql(query, conn)

# 3. Revisar el tamaño del dataset
print(f"Tamaño del dataset: {df.shape[0]} filas y {df.shape[1]} columnas\n")

# 4. Revisar valores nulos por columna
print("--- Nulos por columna ---")
print(df.isnull().sum())

# 5. Revisar filas completamente duplicadas
print(f"\nFilas duplicadas exactas: {df.duplicated().sum()}")
```
